```python
import matplotlib.pyplot as plt
import numpy as np

# Data
models = ['SVM', 'KNN', 'Random Forest']
f1_macro = [0.7668, 0.9840, 0.9891]
f1_micro = [0.9583, 0.9960, 0.9965]
f1_weighted = [0.9579, 0.9960, 0.9965]

# Bar width and positions
bar_width = 0.2
x = np.arange(len(models))

# Plotting
plt.figure(figsize=(10, 6))
plt.bar(x - bar_width, f1_macro, width=bar_width, color='mediumblue', label='F1 Macro')
plt.bar(x, f1_micro, width=bar_width, color='green', label='F1 Micro')
plt.bar(x + bar_width, f1_weighted, width=bar_width, color='indigo', label='F1 Weighted')

# Labels and legend
plt.xlabel('Model', fontsize=10)
plt.ylabel('F1 Score', fontsize=10)
plt.title('Average Metrics Across Folds', fontsize=10)
plt.xticks(x, models, fontsize=12)
plt.ylim(0, 1.1)
plt.legend(fontsize=12)

# Grid and show
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
# Salvar a imagem em alta definição
plt.savefig('grafico_hd.png', dpi=300, bbox_inches='tight')  # DPI alto para qualidade HD

# Exibir o gráfico
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Dados para a tabela
data = {
    "Model": ["RF", "SVM", "KNN"],
    "F1_macro (SMOTE)": [0.98, 0.76, 0.89],
    "F1_micro (SMOTE)": [0.99, 0.95, 0.96],
    "F1_weighted (SMOTE)": [0.99, 0.95, 0.96],
    "F1_macro (No SMOTE)": [0.98, 0.33, 0.89],
    "F1_micro (No SMOTE)": [0.99, 0.55, 0.96],
    "F1_weighted (No SMOTE)": [0.99, 0.69, 0.96],
}

# Criar o DataFrame
df = pd.DataFrame(data)

# Configurar o plot da tabela
fig, ax = plt.subplots(figsize=(8, 4))  # Ajustar o tamanho da figura
ax.axis("off")  # Ocultar os eixos
table = ax.table(
    cellText=df.round(2).values,  # Dados da tabela
    colLabels=df.columns,        # Títulos das colunas
    cellLoc="center",            # Centralizar o texto
    loc="center",                # Localização
)

# Ajustar aparência da tabela
table.auto_set_font_size(True)
table.set_fontsize(10)
table.auto_set_column_width(col=list(range(len(df.columns))))

plt.title("Performance Metrics with and without SMOTE", fontsize=10)
plt.show()
```
